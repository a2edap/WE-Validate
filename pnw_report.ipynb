If you want to save the output from the notebook to a PDF, run the following cell after saving the outputs above. 

```python
def notebook_to_html_to_pdf(notebook_path, output_path = None):

    html_file_name = os.path.basename(notebook_path).replace(".ipynb", ".html")
    if output_path is None:

        output_path = notebook_path.replace(".ipynb", ".pdf")
        html_file = notebook_path.replace(".ipynb", ".html")
    else:
        output_dir = os.path.dirname(output_path)
        html_file = os.path.join(output_dir, html_file_name)
    
    if output_path is None:
        output_path = notebook_path.replace(".ipynb", ".pdf")

    # Convert notebook to HTML, excluding tagged cells
  
    subprocess.run(
            [
                "jupyter", "nbconvert",
                "--to", "html", notebook_path,
                "--TagRemovePreprocessor.enabled=True",
                "--TemplateExporter.exclude_input=True",
                "--TagRemovePreprocessor.remove_cell_tags=exclude_from_pdf"
            ],
            check=True
    )
    generated_html = notebook_path.replace(".ipynb", ".html")
    if generated_html != html_file:
        os.rename(generated_html, html_file)
    # Convert HTML to PDF
    HTML(html_file).write_pdf(output_path, stylesheets=["page_style.css"], presentational_hints=True)


## input notebook name and desired output path ##
notebook_to_html_to_pdf("pnw_report.ipynb", "output/reports/pnw_report.pdf")
```
