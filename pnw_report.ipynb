If you want to save the output from the notebook to a .pdf file, run the following cell after saving the outputs above. 

```python
def notebook_to_html_to_pdf(notebook_path, output_path = None):

    html_file = notebook_path.replace(".ipynb", ".html")
    if output_path is None:
        output_path = notebook_path.replace(".ipynb", ".pdf")

    # Convert notebook to HTML, excluding tagged cells
  
    subprocess.run(
            [
                "jupyter", "nbconvert",
                "--to", "html", notebook_path,
                "--TagRemovePreprocessor.enabled=True",
                "--TagRemovePreprocessor.remove_cell_tags=exclude_from_pdf"
            ],
            check=True
    )

    # Convert HTML to PDF
    HTML(html_file).write_pdf(output_path)


## input notebook name and desired output path ##
notebook_to_html_to_pdf("pnw_report.ipynb", "output/reports/pnw_report.pdf")
```
